```python
def hand_and_till_m_statistic(test_tgt, test_probs, weighted=False):
    def auc_helper(truth, probs):
        fpr, tpr, _ = metrics.roc_curve(truth, probs)
        return metrics.auc(fpr, tpr)

    classes = np.unique(test_tgt)
    n_classes = len(classes)
    indicator = skpre.label_binarize(y=test_tgt, classes=classes)
    avg_auc_sum = 0.0

    # comparing class i and class j
    for ij in it.combinations(classes, 2):
        # use sum to act like a logical OR
        ij_indicator = indicator[:, ij].sum(axis=1, dtype=bool)

        # slightly ugly, can't broadcast these as indexes
        # use np.ix_ to save the day
        ij_probs = test_probs[np.ix_(ij_indicator, ij)]
        ij_test_tgt = test_tgt[ij_indicator]

        i, j = ij
        auc_ij = auc_helper(ij_test_tgt == i, ij_probs[:, 0])
        auc_ji = auc_helper(ij_test_tgt == j, ij_probs[:, 1])

        # compared to Hand and Till reference
        # no /2 ... factor is out since it will cancel
        avg_auc_ij = (auc_ij + auc_ji)

        if weighted:
            avg_auc_ij *= ij_indicator.sum() / len(test_tgt)

        avg_auc_sum += avg_auc_ij

    # compared to Hand and Till reference
    # no *2 ... factored out above and they cancel
    m = avg_auc_sum / (n_classes * (n_classes - 1))
    return m
```

```python
from sklearn.model_selection import StratifiedKFold

model = neighbors.KNeighborsClassifier()
skf = StratifiedKFold(n_splits=10, shuffle=True, random_state=21)

scores = []

for train_idx, test_idx in skf.split(iris.data, iris.target):
    X_train, X_test = iris.data[train_idx], iris.data[test_idx]
    y_train, y_test = iris.target[train_idx], iris.target[test_idx]

    # Skip this fold if fewer than 3 classes are present
    if len(np.unique(y_test)) < 3:
        continue

    model.fit(X_train, y_train)
    y_proba = model.predict_proba(X_test)

    score = hand_and_till_m_statistic(y_test, y_proba)
    scores.append(score)

# Plot results
fig, ax = plt.subplots(1, 1, figsize=(3, 3))
sns.swarmplot(scores, orient='v', ax=ax)
ax.set_title('10-Fold H&T Ms')
```

```python
#We had to redefine the function here to correctly map global class indices to local indices within each fold’s subset, ensuring proper indexing of predicted probabilities and preventing out-of-bounds errors during evaluation. Stupid textbook

def hand_and_till_m_statistic(y_true, y_proba, **kwargs):
    _ = kwargs #Supress unused keword arguments (i.e. passed by make_scorer)
    import numpy as np
    import itertools as it
    from sklearn import metrics
    from sklearn.preprocessing import label_binarize

    def auc_helper(truth, probs):
        fpr, tpr, _ = metrics.roc_curve(truth, probs)
        return metrics.auc(fpr, tpr)

    classes = np.unique(y_true)
    n_classes = len(classes)

    # Ensure y_proba 2D
    if y_proba.ndim == 1:
        y_proba = np.vstack([1 - y_proba, y_proba]).T

    avg_auc_sum = 0.0

    for i, j in it.combinations(range(n_classes), 2):
        ij_mask = (y_true == classes[i]) | (y_true == classes[j])
        ij_true = y_true[ij_mask]
        ij_probs = y_proba[ij_mask]

        present_classes = np.unique(ij_true)
        local_i = np.where(present_classes == classes[i])[0][0]
        local_j = np.where(present_classes == classes[j])[0][0]

        probs_pair = ij_probs[:, [local_i, local_j]]

        auc_ij = auc_helper(ij_true == classes[i], probs_pair[:, 0])
        auc_ji = auc_helper(ij_true == classes[j], probs_pair[:, 1])
        avg_auc_sum += auc_ij + auc_ji

    m = avg_auc_sum / (n_classes * (n_classes - 1))
    return m
```

```python
htm_scorer = metrics.make_scorer(hand_and_till_m_statistic, needs_proba=True)

cv_auc = skms.cross_val_score(model,
                              student_ftrs,
                              student_tgt,
                              scoring=htm_scorer,
                              cv=10)

fig, ax = plt.subplots(figsize=(3, 3))
sns.swarmplot(y=cv_auc, orient='v', ax=ax)
ax.set_title('10-Fold H&T Ms')
```
